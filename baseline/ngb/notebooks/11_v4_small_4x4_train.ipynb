# NGB v4 — small 4-layer / 4-head training

Runs or resumes the distinct two-epoch 4×4 small-language-model protocol. Results live under `/tmp/rg-ngb/results/ngb_v4_small_4x4_2epoch`.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display

cwd = Path.cwd().resolve()
candidates = [cwd, cwd.parent, cwd / 'baseline' / 'ngb']
EXPERIMENT_ROOT = next((p for p in candidates if (p / 'configs' / 'v4_small_4x4.yaml').is_file()), None)
if EXPERIMENT_ROOT is None:
    raise FileNotFoundError('Run from baseline/ngb, baseline/ngb/notebooks, or the repository root')
sys.path.insert(0, str(EXPERIMENT_ROOT / 'src'))

from rg_ngb import canonical_seeds, load_config, roots, run_all_replicates, run_status_table

pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
CONFIG = load_config(EXPERIMENT_ROOT / 'configs' / 'v4_small_4x4.yaml')
SEEDS = canonical_seeds(CONFIG)
PATHS = roots(CONFIG)
display(pd.DataFrame([CONFIG['model'] | CONFIG['training']]))
display(run_status_table(PATHS['results'], seeds=SEEDS))
```

```python
RUNS = run_all_replicates(cfg=CONFIG, seeds=SEEDS, device='auto', resume=True, overwrite=False, progress=True)
print('completed or verified runs:', len(RUNS))
display(run_status_table(PATHS['results'], seeds=SEEDS))
```
