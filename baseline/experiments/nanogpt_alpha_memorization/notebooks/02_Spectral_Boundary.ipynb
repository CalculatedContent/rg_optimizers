# All-matrix alpha boundaries and the guarded intervention

This notebook only reads saved measurements. Error bars use training seeds, not layers or canaries. Partial studies and unsupported fits remain labeled.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Markdown, Image
CODE = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "am_report.py").is_file())
if str(CODE) not in sys.path:
    sys.path.insert(0, str(CODE))
if "STUDY_ROOT" not in globals():
    STUDY_ROOT = Path("/tmp/nanogpt_alpha_memorization_latest.txt").read_text().strip()
ROOT = Path(STUDY_ROOT)
from am_report import generate
REPORT = generate(ROOT)
display(Markdown((REPORT / "summary.md").read_text()))
```

```python
def read_table(name):
    try:
        return pd.read_csv(REPORT / "tables" / name)
    except (FileNotFoundError, pd.errors.EmptyDataError):
        return pd.DataFrame()
display(read_table("regimes.csv"))
spectra = read_table("spectral.csv")
display(spectra[[k for k in ["arm", "seed", "step", "matrix", "alpha_clip_xmax", "alpha_raw", "D", "num_pl_spikes", "rand_distance", "fit_supported"] if k in spectra]])
for image in sorted((REPORT / "figures").glob("alpha_*.png")):
    display(Image(filename=str(image)))
try:
    controls = pd.read_csv(REPORT / "tables/control_events.csv")
    display(controls.groupby(["arm", "seed", "accepted"]).size().rename("windows"))
except pd.errors.EmptyDataError:
    display(Markdown("No control events recorded yet."))
```
