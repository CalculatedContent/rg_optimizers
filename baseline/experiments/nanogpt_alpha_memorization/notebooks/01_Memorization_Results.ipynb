# Memorization and generalization: AdamW vs Muon

This notebook reads saved behavioral measurements from the no-online-WeightWatcher campaign. Error bars use independent training seeds, not canaries. Checkpoints are retained for later post-hoc spectral analysis.

```python
from pathlib import Path
import sys
import pandas as pd
from IPython.display import display, Markdown, Image
CODE = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / "am_report_fast.py").is_file())
if str(CODE) not in sys.path:
    sys.path.insert(0, str(CODE))
if "STUDY_ROOT" not in globals():
    STUDY_ROOT = Path("/tmp/nanogpt_alpha_memorization_latest.txt").read_text().strip()
ROOT = Path(STUDY_ROOT)
from am_report_fast import generate
REPORT = generate(ROOT)
display(Markdown((REPORT / "summary.md").read_text()))
```

```python
display(pd.read_csv(REPORT / "tables/runs.csv"))
for filename in ["final_complete_seed_summary.csv", "final_paired_differences.csv", "compression_summary.csv", "exposure_summary.csv"]:
    try:
        table = pd.read_csv(REPORT / "tables" / filename)
    except (FileNotFoundError, pd.errors.EmptyDataError):
        display(Markdown(f"**{filename}: not yet available.**"))
        continue
    display(Markdown(f"## {filename}"))
    display(table)
for pattern in ["long_64_exact_match.png", "validation_clean_-1_nll.png", "exposure_64.png"]:
    image = REPORT / "figures" / pattern
    if image.exists(): display(Image(filename=str(image)))
```
