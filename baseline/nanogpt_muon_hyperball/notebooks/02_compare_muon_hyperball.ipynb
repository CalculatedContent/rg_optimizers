# Compare long Muon with Muon-HyperBall

Plots task metrics, per-layer alpha and fit distance, ERG gap, traps, and HyperBall activation diagnostics. Incomplete runs can be inspected by setting `REQUIRE_COMPLETE = False`.


```python
from pathlib import Path
import matplotlib.pyplot as plt
import pandas as pd

from rg_nanogpt_muon_hyperball.analysis import (
    load_epoch_metrics,
    load_layer_metrics,
    plot_epoch_metric,
    plot_layer_metric,
    run_status_table,
)

RESULTS_ROOT = Path('/tmp/rg-nanogpt-muon-hyperball/results')
OPTIMIZERS = ('muon', 'muon_hyperball')
SEEDS = (1337,)
REQUIRE_COMPLETE = False

run_status_table(RESULTS_ROOT, optimizers=OPTIMIZERS, seeds=SEEDS)
```

```python
epochs = load_epoch_metrics(
    RESULTS_ROOT,
    optimizers=OPTIMIZERS,
    seeds=SEEDS,
    require_complete=REQUIRE_COMPLETE,
)
layers = load_layer_metrics(
    RESULTS_ROOT,
    optimizers=OPTIMIZERS,
    seeds=SEEDS,
    require_complete=REQUIRE_COMPLETE,
)
epochs.tail(), layers.tail()
```

```python
for metric in ('val_loss', 'val_accuracy', 'test_accuracy', 'primary_lr'):
    if not epochs.empty and metric in epochs.columns:
        plot_epoch_metric(
            epochs,
            metric=metric,
            optimizers=OPTIMIZERS,
            title=f'{metric}: long Muon versus Muon-HyperBall',
        )
        plt.show()
```

```python
for optimizer in OPTIMIZERS:
    for metric in ('alpha', 'D', 'ERG_gap', 'num_traps'):
        if not layers.empty and metric in layers.columns:
            plot_layer_metric(
                layers,
                optimizer=optimizer,
                metric=metric,
            )
            plt.show()
```

```python
hyperball_metrics = (
    'hyperball_active_fraction',
    'hyperball_mean_scale',
    'hyperball_min_scale',
    'hyperball_max_proposed_update_to_weight_ratio',
    'hyperball_max_applied_update_to_weight_ratio',
)
for metric in hyperball_metrics:
    if not epochs.empty and metric in epochs.columns:
        plot_epoch_metric(
            epochs,
            metric=metric,
            optimizers=('muon_hyperball',),
            title=metric,
        )
        plt.show()
```
